### Read

```python
scores = [14.7, 19, 19, 20, 16.8]
# index     0    1   2   3    4
#                    -3  -2   -1
print(scores)
print(scores[3])
print(scores[-2])
print(scores[5])
```

```python
# slicing
scores = [14.7, 19, 19, 20, 16.8]
# index     0    1   2   3    4
# start =====> 0
# stop =======> end of list
# step =======> 1
part1 = scores[:3]
part2 = scores[::3] # 0 , 0 + 3 = 3
list2 = scores[::-1]
print(part2, list2)
print(scores[0:1])
print(scores[0])
print(scores[3:10])
```

### Remove

```python
scores = [14.7, 19, 19, 20, 16.8]
# index     0    1   2   3    4 
# value , index 
```

```python
# remove by value
scores = [14.7, 19, 19, 20, 16.8, 20]
# index     0    1   2   3    4 
scores.remove(20)
print(scores)
# 10000 ====> 20 ====> 1 remove (1)
# 10000 ====> 20 ====> 1 remove =====> 10000
# scores.remove(200)  ## ValueError
print(scores)
```

```python
# remove by index
scores = [14.7, 19, 19, 20, 16.8, 100]
# index     0    1   2   3    4    5
scores.pop(4)
print(scores)
scores.pop()  # -1
print(scores)
# scores.pop(10) #IndexError
```

```python
scores = [14.7, 19, 19, 20, 16.8, 100]
# index     0    1   2   3    4    5
x = scores.remove(20)
print(x)
x = scores.pop(0)
print(x)
```

#### Copy
1. Shallow copy
2. Deep copy

```python
number1 = 10 
number2 = number1
print(number1, number2)
number1 = 30
print(number1, number2)
```

```python
list1 = [10, 2, 40]
list2 = list1 
print(f"before edit: list1 {list1}, list2 {list2}")
list1[0] = 300
list1.append(700)
print(f"after edit: list1 {list1}, list2 {list2}")
```

#### Shallow copy

```python
list1 = [10, 2, 40]
list2 = list1[:] 
print(f"before edit: list1 {list1}, list2 {list2}")
list1[0] = 300
print(f"after edit: list1 {list1}, list2 {list2}")
```

```python
list1 = [10, 2, 40]
list2 = list(list1)
print(f"before edit: list1 {list1}, list2 {list2}")
list1[0] = 300
print(f"after edit: list1 {list1}, list2 {list2}")
```

```python
list1 = [10, 2, 40]
list2 = list1.copy()
print(f"before edit: list1 {list1}, list2 {list2}")
list1[0] = 300
print(f"after edit: list1 {list1}, list2 {list2}")
```

```python
import copy 

list1 = [10, 2, 40]
list2 = copy.copy(list1)
print(f"before edit: list1 {list1}, list2 {list2}")
list1[0] = 300
print(f"after edit: list1 {list1}, list2 {list2}")
```

```python
list1 = [[2, 10], [4, 90]]
list2 = copy.copy(list1)
print(f"before edit: list1 {list1}, list2 {list2}")
list1[0][0] = 300
print(f"after edit: list1 {list1}, list2 {list2}")
```

#### Deep Copy

```python
list1 = [[2, 10], [4, 90]]
list2 = copy.deepcopy(list1)
print(f"before edit: list1 {list1}, list2 {list2}")
list1[0][0] = 300
print(f"after edit: list1 {list1}, list2 {list2}")
```
