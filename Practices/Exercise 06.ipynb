 **Personal Book Manager**
**Write a program that manages your personal library. The program should allow you to store information about books, including whether they have been read or not.**

**Each book should have the following information:**

**-Book title (string)**

**-Author name (string)**

**-Read status (True/False)**

**Limitation: Book titles must be unique.**


**Personal Book Manager**
**The program must support:**

**Adding a new book (title, author, read status)**

**Searching for a book by title**

**Searching for books by author name**

**Showing all books**

**Showing read books**

**Showing unread books**

**Showing a summary:**

**Total number of books**

**Number of books read**

**Number of books unread**

```python
titles = []
authors = []
read_statuses = []
```

```python
def input_data():
    title = input("Enter book title: ").strip()
    author = input("Enter author name: ").strip()
    status_input = input("Has the book been read?(yes/no): ").strip().lower()
    read = True if status_input == 'yes' else False
    return title,author,read
```

```python
title, author,read = input_data()
print(f"Title: {title}, Author: {author}, Read: {read}")
```

```python
def add():
    title, author,read = input_data()
    if title in titles:
        print("This book title alrady exist!")
        return
    titles.append(title)
    authors.append(author)
    read_statuses.append(read)
    print(f"Book added: {title} by {author} | Read: {'Yes' if read else 'No'}")
```

```python
add()
print(titles)
print(authors)
print(read_statuses)
```

```python
def search_by_title():
    title = input("Enter title to search: ").strip()
    if title in titles:
        idx = titles.index(title)
        print(f"{titles[idx]} by {authors[idx]} | Read: {'Yes' if read_statuses[idx] else 'No'}")
    else:
        print("Book not found.")
```

```python
def search_by_author():
    author = input("Enter author to search: ").strip()
    if author in authors:
        idx = authors.index(author)
        print(f"{titles[idx]} by {authors[idx]} | Read: {'Yes' if read_statuses[idx] else 'No'}")
    else:
        print("Book not found.")
```

```python
def search():
    mode = input("Search by 'title' or 'author'? ").strip().lower()
    if mode == "title":
        search_by_title()
    elif mode == "author":
        search_by_author()
    else:
        print("Invalid option. Please type 'title' or 'author'.")
```

```python
titles = ["The midnight library","Atomic habits"]
authors = ["Matt", "James Clear"]
read_statuses= [True,False]
```

```python
search()
```

```python
def show(filter_type):
    if filter_type == "all":
        for i in range(len(titles)):
            print(f"{titles[i]} by {authors[i]} | Read: {'Yes' if read_statuses[i] else 'No'}")
    elif filter_type == "read":
        for i in range(len(titles)):
            if read_statuses[i]:
                print(f"{titles[i]} by {authors[i]} | Read: Yes ")
    elif filter_type == "unread":
         for i in range(len(titles)):
            if not read_statuses[i]:
                print(f"{titles[i]} by {authors[i]} | Read: No ")
```

```python
show("all")
```

```python
show("read")
```

```python
show("unread")
```

```python
def summary():
    total_books = len(titles)
    read_books = sum(read_statuses)
    unread_books = total_books - read_books
    print("Summary: ")
    print(f"Total books: {total_books}")
    print(f"Read books: {read_books}")
    print(f"Unread books: {unread_books}")
```

```python
summary()
```

```python
def help():
    print("Available Commands: ")
    print("add : Add a new book")
    print("search : Search by title or author ")
    print("show all: Show all books")
    print("show read: Show read books")
    print("show unread: Show unread books")
    print("summary : Show statistics ")
    print("exit : Quit program")
```

```python
help()
```

```python
while True:
    command = input("Choose an option (or type 'help'): ").strip().lower()

    if command == "add":
        add()
    elif command == "search":
        search()
    elif command == "show all":
        show("all")
    elif command == "show read":
        show("read")
    elif command == "show unread":
        show("unread")
    elif command == "summary":
        summary()
    elif command == "help":
        help()
    elif command == "exit":
        break
    else:
        print(f"'{command}': Not found")
```
