 **School Management System**
 
 **• Design a Python program to manage information about different school classes and their students.**
 ****
**<span style="color:#FF4C29; font-weight:bold">Data Structure:</span>**

 **• The school consists of multiple classes.**
 
 **• Each class has a unique name.**
 
 **• Each class contains several students.**

**<span style="color:#FA4C29; font-weight:bold">Each student has the following information:</span>** 
 
 **• First Name**
 **• Last Name**
 
 **• GPA**
 
 **• Student ID (automatically generated by the program and unique)** 
 
**<span style="color:#FF4C29; font-weight:bold"> Functionalities:</span>**

 **- add class**

**- add student**

**- remove student**

**- edit student**

**- search student**

**- display all classes and students**

**- summary of total classes and students**

```python
school = {} 
state ={
    "student_counter": 1
}
```

```python
def generate_student_id(state):
    sid = f"S{state['student_counter']:04d}" #S0001
    return sid
```

```python
def ensure_class_exists(class_name):
    return class_name in school
```

```python
def ensure_student_exists(class_name,student_id):
    return student_id in school[class_name]["students"]
```

```python
def add_class():
    class_name = input("Enter class name: ")
    if ensure_class_exists(class_name):
        print("Class already exists.")
        return
    school[class_name] = {"students":{}}
    print("Class added.")
```

```python
def add_student():
    class_name = input("Enter class name: ")
    if not ensure_class_exists(class_name):
        print("Class not found.")
        return
    first_name = input("First name: ")
    last_name = input("Last name: ")
    try: 
        gpa = float(input("GPA: "))
    except ValueError:
        print("Invalid GPA.")
        return
    sid = generate_student_id(state)
    school[class_name]["students"][sid] = {
        "first_name":first_name,
        "last_name":last_name,
        "gpa":gpa
    }
    print(f"Student added with ID {sid}.")
```

```python
def remove_student():
    class_name = input("Enter class name: ")
    if not ensure_class_exists(class_name):
        print("Class not found.")
        return
    student_id = input("Enter student ID: ")
    if not ensure_student_exists(class_name,student_id):
        print("Student not found.")
        return
    del school[class_name]["students"][student_id]
    print("student removed.")
    
```

```python
def edit_student():
    class_name = input("Enter class name: ")
    if not ensure_class_exists(class_name):
        print("Class not found.")
        return
    student_id = input("Enter student ID: ")
    if not ensure_student_exists(class_name,student_id):
        print("Student not found.")
        return
    first_name = input("New first name: ")
    last_name = input("New last name: ")
    try: 
        gpa = float(input("New GPA: "))
    except ValueError:
        print("Invalid GPA.")
        return
    school[class_name]["students"][student_id]= {
        "first_name":first_name,
        "last_name":last_name,
        "gpa":gpa
    }
    print("Student Updated.")
```

```python
def search_student():
    class_name = input("Enter class name: ")
    if not ensure_class_exists(class_name):
        print("Class not found.")
        return
    student_id = input("Enter student ID: ")
    if not ensure_student_exists(class_name,student_id):
        print("Student not found.")
        return
    student = school[class_name]["students"][student_id]
    print(f"Student ID: {student_id}")
    print(f"Name: {student['first_name']} {student['last_name']}")
    print(f"GPA: {student['gpa']}")
```

```python
def display_school():
    if not school:
        print("No classes available.")
        return
    for class_name, class_data in school.items():
        print(f"Class: {class_name}")
        if not class_data["students"]:
            print("No students.")
            continue
        for sid,student in class_data["students"].items():
            print(f"{sid}: {student['first_name']} {student['last_name']}, GPA:{student['gpa']}")
            
```

```python
def summary():
    total_classes = len(school)
    total_students = sum(len(cls["students"]) for cls in school.values())
    print(f"Total classes: {total_classes}")
    print(f"Total students: {total_students}")
```

```python
def help_manu():
    print("Commands: ")
    print("add class    - Add a new class")
    print("add student    - Add a student to a class")
    print("remove student    - Remove a student by ID")
    print("edit student    - Edit student information")
    print("search student    - Search for a student by ID")
    print("display    - Show all classes and students")
    print("summary    - Show total number of classes and students")
    print("exit     - Exit the program")
```

```python
while True:
    action = input("Enter command or 'help': ").strip().lower()
    if action == "add class":
        add_class()
    elif action == "add student":
        add_student()
    elif action == "remove student":
        remove_student()
    elif action == "edit student":
        edit_student()
    elif action == "search student":
        search_student()
    elif action == "display":
        display_school()
    elif action == "summary":
        summary()
    elif action == "help":
        help_manu()
    elif action == "exit":
        break
    elif action == "":
        continue
    else:
        print("Unknown command.")
```
