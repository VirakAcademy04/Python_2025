### Session09 (While)

```python
for i in range(1, 5, 2):
    print(i, end=", ")
```

```python
i = 0 
while i < 5: # 0 < 5 ====> True
    print(i, end=", ")
    i += 1
```

```python
i = 0 
while i < 10:
    number = int(input("Your number: "))
    print(number)
    if number % 2 == 0:
        i += 2 
    else:
        i += 3
```
