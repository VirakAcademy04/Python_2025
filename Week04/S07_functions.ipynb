#### Higher-Order Function

```python
def apply_discount(price):
    return price * 0.95
```

```python
def apply_increase(price):
    return price * 1.03
```

```python
def update_price1(products):
    updated_products = []
    for p in products:
        new_price = apply_discount(p)
        updated_products.append(round(new_price, 3))
    return updated_products
```

```python
def update_price2(products):
    updated_products = []
    for p in products:
        new_price = apply_increase(p)
        updated_products.append(round(new_price, 3))
    return updated_products
```

```python
products = [120, 300, 400, 500, 10]
products1 = update_price1(products)
print(products1)
products2 = update_price2(products)
print(products2)
```

```python
def update_price3(products, update_function):
    updated_products = []
    for p in products:
        new_price = update_function(p)
        updated_products.append(round(new_price, 3))
    return updated_products
```

```python
products = [100, 200, 400]
products1 = update_price3(products, apply_discount)
print(f"After Discount: {products1}")
```

```python
products = [100, 200, 400]
products1 = update_price3(products, apply_increase)
print(f"After Increase: {products1}")
```

### map

```python
list1 = ["12", "34", "100", "20"]
list2 = []
for i in list1:
    list2.append(int(i))
print(list2, type(list2[0]))
```

```python
list1 = ["12", "34", "100", "20"]
list2 = list(map(int, list1))
print(list2, type(list2[0]))
```

```python
list1 = [12, 34, 50, 10, 8] # 0 - 4
list2 = [34, 10, 34, 10]
list3 = []
for i in range(len(list1)):
    result = list1[i]+list2[i]
    list3.append(result)
print(list3)
```

```python
list1 = [12, 34, 50, 10, 8] # 0 - 4
list2 = [34, 10, 34, 10]
list3 = list(map(lambda x, y: x+y, list1, list2))
print(list3)
```

### all

```python
list1 = [True, True, True, True]
result = all(list1)
print(result)
```

```python
list1 = [2, 10, 0, "python"]
result = all(list1)
print(result)
```

```python
list1 = [12, 4, 23, 10, 200]
result = list(map(lambda x: x > 0, list1))
result = all(result)
print(result)
```

```python
list1 = [12, 4, 23, 10, 200]
result = all(map(lambda x: x > 10, list1))
print(result)
```
