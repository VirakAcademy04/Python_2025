#### Read

```python
book = {
    "name":"python", 
    "year":2020, 
    "price":2000
}
book_name = book["name"]
print(book_name)
print(book["year"])
print(book["shop"])
```

```python
book = {
    "name":"python", 
    "year":2020, 
    "price":2000
}
name = book.get("name")
print(name)
shop = book.get("shop")
print(shop)
shop = book.get("shop", "Not Found")
print(shop)
```

#### Search

```python
book = {
    "name":"python", 
    "year":2020,
    "page": 3000
}
# in, not in 
# key: name, year, page 
# value: python, 2020, 3000
# item: (name, python), (year, 2020), (page, 3000)
```

```python
book = {
    "name":"python", 
    "year":2020,
    "page": 3000
}
book_keys = book.keys()
print(book_keys)
book_values = book.values()
print(book_values)
book_items = book.items()
print(book_items)
```

```python
book = {
    "name":"python", 
    "year":2020,
    "page": 3000
}
if "name" in book.keys():
    print(book["name"])

if 2020 in book.values():
    print("yes")

if ("year", 2020) in book.items():
    print("Yes")

if "name" in book:
    print(book["name"])
```

#### Loop

```python
book = {
    "name":"python", 
    "year":2020,
    "page": 3000
}
for i in book: 
    print(i)
```

```python
book = {
    "name":"python", 
    "year":2020,
    "page": 3000
}
for i in book.values(): 
    print(i)
```

```python
book = {
    "name":"python", 
    "year":2020,
    "page": 3000
}
for i, data in book.items(): 
    print(data)
```
