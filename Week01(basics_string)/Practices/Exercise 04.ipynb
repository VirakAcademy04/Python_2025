### <span style="color:#FF4C29; font-weight:bold">Question01</span>
**Write a programm that recieves the name and price of five products from the user. The program should perform the following tasks:**
- For each products, display its name and price.
- If the price of a product is more than 500, lable it as "Expensive", otherwise lable it as "Cheap".
- Calculate and display the average price of all products.
- Identify and display the name and price of the cheapset product.
- Identify and display the name and price of the most expensive product.


```python
total_price = 0
name = input("Enter name of product 1: ")
price = float(input("Enter price of product 1: "))
total_price += price #total_price = total_price + price

print(f"Product: {name}, Price: {price}")
print("Label: ", "Expensive" if price > 500 else "Cheap")

min_price = price
max_price = price
min_name = name
max_name = name

for i in range(2,6):
    name = input(f"Enter name of product {i}: ")
    price = float(input(f"Enter price of product {i}: "))
    total_price += price

    print(f"Product: {name}, Price: {price}")
    print("Label: ", "Expensive" if price > 500 else "Cheap")

    if price < min_price:
        min_price = price
        min_name = name
    
    if price > max_price :
        max_price = price
        max_name = name

average_price = total_price / 5
print("Average price of all products: ", average_price)

print("Cheapest product: ", min_name, "with price: ", min_price)
print("Most expensive product: ", max_name, "with price: ", max_price)
```

### <span style="color:#FF4C29; font-weight:bold">Question02</span>
 **Write a program that takes a number from the user and calculates its factorial**

```python
# 4! = 1 * 2 * 3 * 4 * 5
number = int(input("Enter a number: "))
factorial = 1
if number < 0 :
    print("Factorial does not exist for negative numbers. ")
elif number == 0:
    print("The Facorial of 0 is 1.")
else:
    for i in range(1,number + 1):
        factorial *= i
    print(f"The factorial of {number} is {factorial}")
```

### <span style="color:#FF4C29; font-weight:bold">Question03</span>
**Write a program that takes the names of five different files from the user and display how many of them have the .py extension.**

```python
py_count = 0

for i in range(5):
    filename = input (f"Enter the name of file {i + 1}: ")
    if filename.endswith(".py"):
        py_count += 1
print(f"{py_count} of the files have a .py extension.")
```
