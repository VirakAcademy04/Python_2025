## Session13 (Set)


    Unordered collection

    Unique elements only

    No indexing or slicing

    Removes duplicates automatically

```python
# [] ====> index, value
# {} ====> key:value
# () ====> index, value 
# {}
```

```python
data = {12, 4, 20, 30}
print(data)
print(type(data))
print(len(data))
```

### Define

```python
data = {12, 30, 0, 0.0, False, True, 30}
print(data)
print(len(data))
```

```python
set1 = set()
print(type(set1), len(set1))
```

```python
set1 = {}
print(type(set1), len(set1))
```

### Read data

```python
data = {12, 10, 200, 300, 12, 3}
print(data)
for i in data:
    print(i, end=", ")
```

### Add

```python
data = {12, 3, 14, 50, 12}
data.add(20)
data.add(50)
print(data)
```

```python
data = {34, 20, 50}
data.update([30, 400, 50])
print(len(data))
print(data)
```

### Remove

```python
data = {34, 20, 50, 20, 10}
data.remove(34)
print(data)
data.remove(100)
```

```python
data = {34, 20, 50, 20, 10}
data.discard(34)
print(data)
data.discard(100)
```

```python
data = {34, 20, 50, 20, 10}
data.clear()
print(data)
```

```python
data = {34, 20, 50, 20, 10}
print(data)
result = data.pop()
print(data)
print(result)
```

### Search

```python
# in, not in 
data = {34, 20, 50, 20, 10}
if 50 in data:
    print("yes")
```
