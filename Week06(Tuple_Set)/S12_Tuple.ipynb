### NamedTuple

```python
student = ("Diana", 15, 19)
print(student[2])
```

```python
from collections import namedtuple

Student = namedtuple("Student", ["name", "age", "score"])
st1 = Student(name="Sara", age=18, score=20)
print(st1)
print(st1.name)
```

```python
Product = namedtuple("Product", ["name", "price", "number"])
p1 = Product("book", 3000, 2)
p2 = Product("phone", 20000, 1)
products = [p1, p2]

total_value = sum(p.price * p.number for p in products)
print(total_value)
```

```python
prices = []
for p in products:
    prices.append(p.price * p.number)
print(prices)
print(sum(prices))
```

```python
# dict
user1 = {"name": "Sara", "age": 18}
user2 = {"usename": "Ali", "years": 19}
print(user1, user2)

# namedtuple
User = namedtuple("User", ["name", "age"])
user1 = User("Zahra", 16)
user2 = User("Diana", 18)
print(user1, user2)
```
